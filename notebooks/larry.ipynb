# LARRY

A lineage-traced hematopoietic differentiation dataset (LARRY, Weinreb et al. (2020)), with experimentally observed clonal fates as references. The source $p_0$ is the 4,638 Day-2 cells and the target $p_1$ a subsample of 8,000 Day-6 cells, in the first $d$ principal components; RNA velocity and PCA are fit on these two days, and the models do not use Day 4. For each seed, the clones are split in half: lineage enters the transition kernel only through the training clones, and the Day-2 cells of held-out clones with at least five differentiated Day-4 or Day-6 relatives are scored. Hyperparameters are transferred without retuning as the log-median of the tuned Pancreas, Erythroid and iTracer points; the noise width of the models with noise is the transferred 0.0547723 at $d = 50$ and 0.3 at $d = 20$, the width chosen on the Pancreas fate cells.

## Endpoint $W_2$

Endpoint $W_2$ compares the $t = 1$ states of the scored cells, one trajectory per cell, with 2,000 cells drawn at random from the Day-6 target; mean ± population SD over five seeds. The fate runs record only trajectory endpoints, so no midpoint is reported; no sampling floor is recorded, and the constant predictor has no trajectories.

## Fate prediction (Table 2, LARRY)

FFM uses a velocity-derived transition kernel, augmented with lineage in the last row; each trajectory endpoint takes the most frequent fate among its 15 nearest differentiated Day-6 cells, and a source cell's fate probabilities are the fractions over its 64 sampled trajectories for the models with noise (one trajectory for deterministic FFM). We report centered MSE (cMSE) between predicted and reference fate probabilities, which removes differences in average fate composition; values are mean ± population SD over five seeds. The constant predictor is the mean clonal fate of the training-clone Day-2 cells with at least five differentiated relatives; like any constant prediction, its cMSE equals the total variance of the reference fates over the scored cells.
