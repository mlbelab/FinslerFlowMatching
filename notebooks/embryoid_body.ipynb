# Embryoid Body

The Embryoid Body (EB) time course (Moon et al., 2019) has five time points, days 0–4, represented by the first $d$ principal components. We follow the leave-one-time-point-out (LOTO) protocol, withholding each of days 1–3 in turn: every method is trained on days $k-1$ and $k+1$, and its state at $t = 1/2$ is compared with the withheld day $k$. FFM uses an expression-derived kernel built from the two training days; hyperparameters are transferred without retuning from the pseudotime benchmarks, except that Curly-FM uses $\alpha = 1$ with the velocity rescaled to the median source–target displacement. The principal components and the Curly-FM velocity are those of the TrajectoryNet release, computed on all cells including the withheld day; the velocity cannot be refit without spliced/unspliced counts, which the release does not contain.
